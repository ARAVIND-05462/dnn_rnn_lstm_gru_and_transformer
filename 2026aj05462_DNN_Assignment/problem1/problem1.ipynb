##### Inspect the HAR data

```python
dataset_path = "data/har/UCI HAR Dataset"

channel_names = [
    "body_acc_x", "body_acc_y", "body_acc_z",
    "body_gyro_x", "body_gyro_y", "body_gyro_z",
    "total_acc_x", "total_acc_y", "total_acc_z"
]

def load_har_split(split):
    split_path = os.path.join(dataset_path, split, "Inertial Signals")

    signals = []

    for channel in channel_names:
        file_path = os.path.join(
            split_path,
            f"{channel}_{split}.txt"
        )

        channel_data = np.loadtxt(file_path)
        signals.append(channel_data)

    # Each channel has shape (number of samples, 128).
    # Stack channels along the last axis.
    X = np.stack(signals, axis=-1)

    y = np.loadtxt(
        os.path.join(dataset_path, split, f"y_{split}.txt")
    ).astype(np.int64)

    subjects = np.loadtxt(
        os.path.join(dataset_path, split, f"subject_{split}.txt")
    ).astype(np.int64)

    return X.astype(np.float32), y, subjects


X_train, y_train, subject_train = load_har_split("train")
X_test, y_test, subject_test = load_har_split("test")

print("Training data shape:", X_train.shape)
print("Training labels shape:", y_train.shape)
print("Training subjects shape:", subject_train.shape)

print("Test data shape:", X_test.shape)
print("Test labels shape:", y_test.shape)
print("Test subjects shape:", subject_test.shape)

print("Unique activity labels:", np.unique(y_train))
print("Unique training subjects:", np.unique(subject_train))
```

#### Split training subjects into training and validation groups.


```python
def split_subjects(subject_ids, seed):
    unique_subjects = np.unique(subject_ids)

    rng = np.random.default_rng(seed)

    validation_subjects = rng.choice(
        unique_subjects,
        size=4,
        replace=False
    )

    training_subjects = np.setdiff1d(
        unique_subjects,
        validation_subjects
    )

    return training_subjects, validation_subjects


# Temporary seed for checking the pipeline.
split_seed = 42

train_subjects, val_subjects = split_subjects(
    subject_train,
    split_seed
)

# Select windows based on subject membership.
train_mask = np.isin(subject_train, train_subjects)
val_mask = np.isin(subject_train, val_subjects)

X_train_split = X_train[train_mask]
y_train_split = y_train[train_mask]

X_val = X_train[val_mask]
y_val = y_train[val_mask]

print("Training subjects:", train_subjects)
print("Validation subjects:", val_subjects)

print("\nTraining data:", X_train_split.shape)
print("Training labels:", y_train_split.shape)

print("\nValidation data:", X_val.shape)
print("Validation labels:", y_val.shape)
```

##### Adding problem generator

```python
def make_adding(n, T, seed):
  """
  Generate the Adding Problem dataset.

  Parameters:
      n    : Number of sequences
      T    : Sequence length
      seed : Random seed

  Returns:
      x : Input sequences of shape (n, T, 2)
      y : Target sums of shape (n,)
  """
  rng = np.random.default_rng(seed)

  # Channel 0 contains random values.
  # Channel 1 contains two marker positions.
  x = np.zeros((n, T, 2), dtype=np.float32)

  # Generate random values in channel 0
  x[:, :, 0] = rng.random((n, T))

  # Select one position from each half
  i1 = rng.integers(0, T // 2, n)
  i2 = rng.integers(T // 2, T, n)

  # Place the two markers in channel 1
  x[np.arange(n), i1, 1] = 1
  x[np.arange(n), i2, 1] = 1

  # Target is the sum of the two marked values
  y = x[np.arange(n), i1, 0] + x[np.arange(n), i2, 0]

  return x, y.astype(np.float32)
```

### 2(a): Implement Vanilla RNN from scratch

```python
class MyRNNCell(torch.nn.Module):
  def __init__(self, input_size, hidden_size):
    super().__init__()
    self.input_size = input_size
    self.hidden_size = hidden_size

    # Input-to-hidden weights: (hidden_size, input_size)
    self.weight_ih = torch.nn.Parameter(
        torch.empty(hidden_size, input_size)
    )

    # Hidden-to-hidden weights: (hidden_size, hidden_size)
    self.weight_hh = torch.nn.Parameter(
        torch.empty(hidden_size, hidden_size)
    )

    # Combined bias: (hidden_size,)
    self.bias = torch.nn.Parameter(
        torch.empty(hidden_size)
    )

    # Initialize parameters
    torch.nn.init.xavier_uniform_(self.weight_ih)
    torch.nn.init.orthogonal_(self.weight_hh)
    torch.nn.init.zeros_(self.bias)

  def forward(self, x_t, h_prev):
    # Input contribution
    input_part = torch.nn.functional.linear(
        x_t, self.weight_ih
    )

    # Previous hidden-state contribution
    hidden_part = torch.nn.functional.linear(
        h_prev, self.weight_hh
    )

    # Combine contributions and apply tanh
    h_t = torch.tanh(
        input_part + hidden_part + self.bias
    )

    return h_t
```

### 2(a): Implement LSTM Cell from scratch

```python

class MyLSTMCell(torch.nn.Module):
  def __init__(self, input_size, hidden_size):
    super().__init__()
    self.input_size = input_size
    self.hidden_size = hidden_size

    # Weight matrices for all four gates: input, forget, candidate, output
    self.weight_ih = torch.nn.Parameter(
        torch.empty(4 * hidden_size, input_size)
    )
    self.weight_hh = torch.nn.Parameter(
        torch.empty(4 * hidden_size, hidden_size)
    )
    self.bias = torch.nn.Parameter(
        torch.empty(4 * hidden_size)
    )

    torch.nn.init.xavier_uniform_(self.weight_ih)
    torch.nn.init.orthogonal_(self.weight_hh)
    torch.nn.init.zeros_(self.bias)

  def forward(self, x_t, h_prev, c_prev):
    # Calculate the input and recurrent contributions
    gates = (
        torch.nn.functional.linear(x_t, self.weight_ih)
        + torch.nn.functional.linear(h_prev, self.weight_hh)
        + self.bias
    )

    # Split the combined result into four gates (PyTorch order: i, f, g, o)
    i_gate, f_gate, g_gate, o_gate = gates.chunk(4, dim=1)

    # Apply the corresponding activation functions
    i_gate = torch.sigmoid(i_gate)
    f_gate = torch.sigmoid(f_gate)
    g_gate = torch.tanh(g_gate)
    o_gate = torch.sigmoid(o_gate)

    # Update the cell state and hidden state
    c_t = f_gate * c_prev + i_gate * g_gate
    h_t = o_gate * torch.tanh(c_t)

    return h_t, c_t
```

### 2(c): Implement GRU Cell from scratch

```python
class MyGRUCell(torch.nn.Module):
  def __init__(self, input_size, hidden_size):
    super().__init__()
    self.input_size = input_size
    self.hidden_size = hidden_size

    # Weight matrices for reset, update and candidate gates
    self.weight_ih = torch.nn.Parameter(
        torch.empty(3 * hidden_size, input_size)
    )
    self.weight_hh = torch.nn.Parameter(
        torch.empty(3 * hidden_size, hidden_size)
    )

    # Separate biases to match PyTorch GRU computation
    self.bias_ih = torch.nn.Parameter(
        torch.empty(3 * hidden_size)
    )
    self.bias_hh = torch.nn.Parameter(
        torch.empty(3 * hidden_size)
    )

    torch.nn.init.xavier_uniform_(self.weight_ih)
    torch.nn.init.orthogonal_(self.weight_hh)
    torch.nn.init.zeros_(self.bias_ih)
    torch.nn.init.zeros_(self.bias_hh)

  def forward(self, x_t, h_prev):
    # Input and recurrent contributions
    input_gates = torch.nn.functional.linear(
        x_t, self.weight_ih, self.bias_ih
    )

    hidden_gates = torch.nn.functional.linear(
        h_prev, self.weight_hh, self.bias_hh
    )

    # Split gates in PyTorch order: reset, update, new
    input_r, input_z, input_n = input_gates.chunk(3, dim=1)
    hidden_r, hidden_z, hidden_n = hidden_gates.chunk(3, dim=1)

    # Reset and update gates
    r_gate = torch.sigmoid(input_r + hidden_r)
    z_gate = torch.sigmoid(input_z + hidden_z)

    # Candidate hidden state
    n_gate = torch.tanh(input_n + r_gate * hidden_n)

    # Final hidden state
    h_t = (1 - z_gate) * n_gate + z_gate * h_prev

    return h_t
```

##### Build the three HAR classifiers

```python
import torch.nn as nn

HIDDEN_SIZE = 64
INPUT_SIZE = 9
NUM_CLASSES = 6

class HARRecurrentClassifier(nn.Module):
  def __init__(self, model_type, input_size=9, hidden_size=64, num_classes=6):
    super().__init__()
    self.model_type = model_type.upper()

    if self.model_type == "RNN":
        self.recurrent = nn.RNN(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True,
            nonlinearity="tanh"
        )

    elif self.model_type == "LSTM":
        self.recurrent = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )

    elif self.model_type == "GRU":
        self.recurrent = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )

    else:
        raise ValueError("Model type must be RNN, LSTM or GRU")

    self.classifier = nn.Linear(hidden_size, num_classes)

  def forward(self, x):
    # Input shape: (batch_size, sequence_length, input_size)
    if self.model_type == "LSTM":
        _, (hidden, cell) = self.recurrent(x)
    else:
        _, hidden = self.recurrent(x)

    # Final hidden state of the last recurrent layer
    final_hidden = hidden[-1]

    # Convert the final representation into class logits
    logits = self.classifier(final_hidden)

    return logits

# Verify all three architectures
for model_name in ["RNN", "LSTM", "GRU"]:
  model = HARRecurrentClassifier(model_name)

  sample_output = model(sample_inputs)

  params = sum(p.numel() for p in model.parameters())

  print(f"\n{model_name}")
  print("Output shape:", sample_output.shape)
  print("Trainable parameters:", params)

  assert sample_output.shape == (128, 6)

print("\nAll three classifier shape checks passed!")
```

##### Shared training and evaluation functions

```python

import time
import copy

from sklearn.metrics import (accuracy_score, f1_score, confusion_matrix)

LEARNING_RATE = 1e-3
EPOCHS = 15

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Training device:", device)

def evaluate_model(model, data_loader, criterion):
  model.eval()

  total_loss = 0.0
  all_predictions = []
  all_labels = []

  with torch.no_grad():
    for inputs, labels in data_loader:
      inputs = inputs.to(device)
      labels = labels.to(device)

      logits = model(inputs)
      loss = criterion(logits, labels)

      total_loss += loss.item() * labels.size(0)

      predictions = torch.argmax(logits, dim=1)

      all_predictions.extend(predictions.cpu().numpy())
      all_labels.extend(labels.cpu().numpy())

  average_loss = total_loss / len(data_loader.dataset)

  accuracy = accuracy_score(all_labels, all_predictions)
  macro_f1 = f1_score(
      all_labels,
      all_predictions,
      average="macro"
  )

  matrix = confusion_matrix(
      all_labels,
      all_predictions,
      labels=list(range(NUM_CLASSES))
  )

  return average_loss, accuracy, macro_f1, matrix


def train_one_model(model, train_loader, val_loader):
  model = model.to(device)
  criterion = nn.CrossEntropyLoss()

  optimizer = torch.optim.Adam(
      model.parameters(),
      lr=LEARNING_RATE
  )

  best_val_f1 = -1.0
  best_epoch = 0
  best_state = None

  epoch_times = []
  history = []

  for epoch in range(EPOCHS):
      model.train()

      running_loss = 0.0

      if device.type == "cuda":
          torch.cuda.synchronize()

      start_time = time.perf_counter()

      for inputs, labels in train_loader:
          inputs = inputs.to(device)
          labels = labels.to(device)

          optimizer.zero_grad(set_to_none=True)

          logits = model(inputs)
          loss = criterion(logits, labels)

          loss.backward()
          optimizer.step()

          running_loss += loss.item() * labels.size(0)

      if device.type == "cuda":
          torch.cuda.synchronize()

      epoch_time = time.perf_counter() - start_time
      epoch_times.append(epoch_time)

      train_loss = running_loss / len(train_loader.dataset)

      val_loss, val_acc, val_f1, _ = evaluate_model(
          model, val_loader, criterion
      )

      history.append({
          "epoch": epoch + 1,
          "train_loss": train_loss,
          "val_loss": val_loss,
          "val_accuracy": val_acc,
          "val_macro_f1": val_f1
      })

      # Keep the checkpoint with the best validation macro-F1
      if val_f1 > best_val_f1:
          best_val_f1 = val_f1
          best_epoch = epoch + 1
          best_state = copy.deepcopy(model.state_dict())

      print(
          f"Epoch [{epoch + 1:02d}/{EPOCHS}] "
          f"Train Loss: {train_loss:.4f} | "
          f"Val Loss: {val_loss:.4f} | "
          f"Val Accuracy: {val_acc:.4f} | "
          f"Val Macro-F1: {val_f1:.4f} | "
          f"Time: {epoch_time:.2f}s"
      )

  # Restore the best validation checkpoint
  model.load_state_dict(best_state)

  return model, history, best_epoch, best_val_f1, np.mean(epoch_times)
```

#### Training and Evaluation

```python
# from torch.utils.data import TensorDataset, DataLoader
def set_adding_seed(seed):
  random.seed(seed)
  np.random.seed(seed)
  torch.manual_seed(seed)

  if torch.cuda.is_available():
      torch.cuda.manual_seed_all(seed)


def create_adding_loaders(data, seed, batch_size=128):
  X_train = torch.tensor(data["X_train"], dtype=torch.float32)
  y_train = torch.tensor(data["y_train"], dtype=torch.float32)

  X_val = torch.tensor(data["X_val"], dtype=torch.float32)
  y_val = torch.tensor(data["y_val"], dtype=torch.float32)

  X_test = torch.tensor(data["X_test"], dtype=torch.float32)
  y_test = torch.tensor(data["y_test"], dtype=torch.float32)

  generator = torch.Generator()
  generator.manual_seed(seed)

  train_loader = DataLoader(
      TensorDataset(X_train, y_train),
      batch_size=batch_size,
      shuffle=True,
      generator=generator
  )

  val_loader = DataLoader(
      TensorDataset(X_val, y_val),
      batch_size=batch_size,
      shuffle=False
  )

  test_loader = DataLoader(
      TensorDataset(X_test, y_test),
      batch_size=batch_size,
      shuffle=False
  )

  return train_loader, val_loader, test_loader


def evaluate_adding_model(model, loader, device):
  model.eval()
  total_squared_error = 0.0
  total_samples = 0

  with torch.no_grad():
    for X_batch, y_batch in loader:
      X_batch = X_batch.to(device)
      y_batch = y_batch.to(device)

      predictions = model(X_batch)
      squared_error = (predictions - y_batch) ** 2

      total_squared_error += squared_error.sum().item()
      total_samples += y_batch.size(0)

  return total_squared_error / total_samples


def train_adding_model(model, train_loader, val_loader, epochs=20, learning_rate=1e-3, device=adding_device):
    model = model.to(device)

    criterion = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

    history = {"train_mse": [], "val_mse": []}

    best_val_mse = float("inf")
    best_state = None
    epoch_times = []

    for epoch in range(epochs):
        model.train()

        if device.type == "cuda":
            torch.cuda.synchronize()

        epoch_start = time.perf_counter()

        for X_batch, y_batch in train_loader:
          X_batch = X_batch.to(device)
          y_batch = y_batch.to(device)

          optimizer.zero_grad()

          predictions = model(X_batch)
          loss = criterion(predictions, y_batch)

          loss.backward()
          optimizer.step()

        if device.type == "cuda":
          torch.cuda.synchronize()

        epoch_times.append(time.perf_counter() - epoch_start)

        train_mse = evaluate_adding_model(
            model, train_loader, device
        )

        val_mse = evaluate_adding_model(
            model, val_loader, device
        )

        history["train_mse"].append(train_mse)
        history["val_mse"].append(val_mse)

        if val_mse < best_val_mse:
          best_val_mse = val_mse
          best_state = copy.deepcopy(model.state_dict())

    model.load_state_dict(best_state)

    return model, history, best_val_mse, np.mean(epoch_times)
```

##### Capture hidden-state gradients

```python
import torch.nn.functional as F
def get_hidden_gradient_norms(model, X_batch, y_batch, device):
    model.eval()
    model.zero_grad(set_to_none=True)

    X_batch = X_batch.to(device)
    y_batch = y_batch.to(device)

    batch_size = X_batch.size(0)
    hidden_size = model.recurrent.hidden_size

    # Initial hidden state
    hidden = torch.zeros(
        batch_size,
        hidden_size,
        device=device
    )

    # Extract the trained PyTorch RNN parameters
    weight_ih = model.recurrent.weight_ih_l0
    weight_hh = model.recurrent.weight_hh_l0
    bias_ih = model.recurrent.bias_ih_l0
    bias_hh = model.recurrent.bias_hh_l0

    hidden_states = []

    # Explicitly unroll the RNN over all 100 timesteps
    for t in range(X_batch.size(1)):
        input_t = X_batch[:, t, :]

        hidden = torch.tanh(
            F.linear(input_t, weight_ih, bias_ih)
            + F.linear(hidden, weight_hh, bias_hh)
        )

        hidden.retain_grad()
        hidden_states.append(hidden)

    # Prediction from the final hidden state
    prediction = model.regressor(
        hidden_states[-1]
    ).squeeze(-1)

    loss = F.mse_loss(prediction, y_batch)

    # Backpropagate through the complete sequence
    loss.backward()

    # Average hidden-state gradient L2 norm across samples
    gradient_norms = []

    for state in hidden_states:
        norm = state.grad.norm(p=2, dim=1).mean().item()
        gradient_norms.append(norm)

    return np.array(gradient_norms), loss.item()


# Use exactly the same 128 examples before and after training
X_grad = torch.tensor(
    adding_data[GRAD_T]["X_train"][:128],
    dtype=torch.float32
)

y_grad = torch.tensor(
    adding_data[GRAD_T]["y_train"][:128],
    dtype=torch.float32
)

# --------------------------------------------------
# 1. Gradients BEFORE training
# --------------------------------------------------
grad_model.load_state_dict(initial_rnn_state)

before_gradients, before_loss = get_hidden_gradient_norms(
    grad_model,
    X_grad,
    y_grad,
    adding_device
)

print("Before training loss:", before_loss)
print("Initial gradient norm:", before_gradients[0])
print("Final gradient norm:", before_gradients[-1])


# --------------------------------------------------
# 2. Train the RNN
# --------------------------------------------------
grad_model, grad_history, grad_best_val, grad_epoch_time = (
    train_adding_model(
        grad_model,
        grad_train_loader,
        grad_val_loader,
        epochs=ADDING_EPOCHS,
        learning_rate=ADDING_LEARNING_RATE,
        device=adding_device
    )
)

print("\nRNN training completed.")
print("Best validation MSE:", grad_best_val)


# --------------------------------------------------
# 3. Gradients AFTER training
# --------------------------------------------------
after_gradients, after_loss = get_hidden_gradient_norms(
    grad_model,
    X_grad,
    y_grad,
    adding_device
)

print("\nAfter training loss:", after_loss)
print("Initial gradient norm:", after_gradients[0])
print("Final gradient norm:", after_gradients[-1])


# --------------------------------------------------
# 4. Plot gradient norms
# --------------------------------------------------
timesteps = np.arange(1, GRAD_T + 1)

plt.figure(figsize=(10, 6))

plt.semilogy(
    timesteps,
    np.maximum(before_gradients, 1e-12),
    label="Before training",
    linewidth=2
)

plt.semilogy(
    timesteps,
    np.maximum(after_gradients, 1e-12),
    label="After training",
    linewidth=2
)

plt.xlabel("Timestep (t)")
plt.ylabel(r"Mean gradient norm $||\partial L/\partial h_t||_2$")
plt.title("RNN Hidden-State Gradient Flow (T=100)")
plt.legend()
plt.grid(True, which="both", linestyle="--", alpha=0.5)

plt.tight_layout()

gradient_figure_path = (
    "problem1/figures/rnn_gradient_flow_T100.png"
)

plt.savefig(gradient_figure_path, dpi=300)
plt.show()

print("Figure saved to:", gradient_figure_path)
```

### Gradient Flow Analysis – Vanilla RNN (T=100)

The hidden-state gradient norms were examined before and after training a vanilla RNN on the adding problem with sequence length 100. The gradient norm at each timestep was calculated by backpropagating the mean squared error loss and averaging the L2 norms across the batch.

Before training, the gradient norm at the first timestep was effectively zero, while the final timestep had a gradient norm of 0.00730. After training, the first-timestep gradient remained effectively zero, and the final-timestep gradient decreased to 0.00267.

This demonstrates the vanishing gradient problem in vanilla RNNs. During backpropagation through time, gradients pass through repeated recurrent transformations and can shrink considerably as they propagate towards earlier timesteps. Consequently, the network struggles to learn dependencies involving information from distant positions in the sequence.

Although training reduced the loss from 0.89032 to 0.15241, the RNN did not learn the adding task effectively. This suggests that optimization alone does not resolve the difficulty of preserving long-range information in a vanilla recurrent architecture.

#### Implement and run the hidden-size sweep

```python
# 1. configuration

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

EPOCHS_ABLATION = 15
BATCH_SIZE_ABLATION = 128
LR_ABLATION = 1e-3

print("Device:", DEVICE)
print("Total experiments:", 2 * len(HIDDEN_SIZES) * len(HAR_SEEDS))

# 2. Reproducibility

def set_har_ablation_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# 3. LSTM / GRU classifier
class HARSizeAblationModel(nn.Module):

    def __init__(self, model_type, hidden_size):
        super().__init__()

        self.model_type = model_type

        if model_type == "LSTM":
            self.recurrent = nn.LSTM(
                input_size=9,
                hidden_size=hidden_size,
                batch_first=True
            )

        elif model_type == "GRU":
            self.recurrent = nn.GRU(
                input_size=9,
                hidden_size=hidden_size,
                batch_first=True
            )

        else:
            raise ValueError("model_type must be LSTM or GRU")

        self.classifier = nn.Linear(hidden_size, 6)

    def forward(self, x):
        output, hidden = self.recurrent(x)

        if self.model_type == "LSTM":
            final_hidden = hidden[0][-1]
        else:
            final_hidden = hidden[-1]

        return self.classifier(final_hidden)

# 4. DataLoader helper

def make_har_ablation_loaders(data, seed):

    Xtr = torch.tensor(data["X_train"], dtype=torch.float32)
    ytr = torch.tensor(data["y_train"] - 1, dtype=torch.long)

    Xval = torch.tensor(data["X_val"], dtype=torch.float32)
    yval = torch.tensor(data["y_val"] - 1, dtype=torch.long)

    Xte = torch.tensor(data["X_test"], dtype=torch.float32)
    yte = torch.tensor(data["y_test"] - 1, dtype=torch.long)

    generator = torch.Generator()
    generator.manual_seed(seed)

    train_loader = DataLoader(
        TensorDataset(Xtr, ytr),
        batch_size=BATCH_SIZE_ABLATION,
        shuffle=True,
        generator=generator
    )

    val_loader = DataLoader(
        TensorDataset(Xval, yval),
        batch_size=BATCH_SIZE_ABLATION,
        shuffle=False
    )

    test_loader = DataLoader(
        TensorDataset(Xte, yte),
        batch_size=BATCH_SIZE_ABLATION,
        shuffle=False
    )

    return train_loader, val_loader, test_loader

# 5. Evaluation helper
def evaluate_har_ablation(model, loader, criterion):

    model.eval()

    total_loss = 0.0
    all_predictions = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in loader:

            X_batch = X_batch.to(DEVICE)
            y_batch = y_batch.to(DEVICE)

            logits = model(X_batch)
            loss = criterion(logits, y_batch)

            total_loss += loss.item() * len(y_batch)

            predictions = logits.argmax(dim=1)

            all_predictions.extend(predictions.cpu().numpy())
            all_targets.extend(y_batch.cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)

    macro_f1 = f1_score(
        all_targets,
        all_predictions,
        labels=np.arange(6),
        average="macro",
        zero_division=0
    )

    accuracy = accuracy_score(all_targets, all_predictions)

    return avg_loss, accuracy, macro_f1

# 6. Train one experiment
def train_har_ablation(model_type, hidden_size, seed):

    set_har_ablation_seed(seed)

    data = har_ablation_data[seed]

    train_loader, val_loader, test_loader = (
        make_har_ablation_loaders(data, seed)
    )

    model = HARSizeAblationModel(
        model_type=model_type,
        hidden_size=hidden_size
    ).to(DEVICE)

    params = sum(
        p.numel() for p in model.parameters()
        if p.requires_grad
    )

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LR_ABLATION
    )

    best_val_f1 = -1.0
    best_epoch = 0
    best_state = None
    epoch_times = []

    for epoch in range(EPOCHS_ABLATION):

        model.train()

        if DEVICE.type == "cuda":
            torch.cuda.synchronize()

        epoch_start = time.perf_counter()

        for X_batch, y_batch in train_loader:

            X_batch = X_batch.to(DEVICE)
            y_batch = y_batch.to(DEVICE)

            optimizer.zero_grad()

            logits = model(X_batch)
            loss = criterion(logits, y_batch)

            loss.backward()
            optimizer.step()

        if DEVICE.type == "cuda":
            torch.cuda.synchronize()

        epoch_time = time.perf_counter() - epoch_start
        epoch_times.append(epoch_time)

        # Select checkpoint using validation macro-F1
        val_loss, val_acc, val_f1 = evaluate_har_ablation(
            model,
            val_loader,
            criterion
        )

        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            best_epoch = epoch + 1
            best_state = copy.deepcopy(model.state_dict())

        print(
            f"Epoch {epoch + 1:02d}/{EPOCHS_ABLATION} | "
            f"Val loss: {val_loss:.4f} | "
            f"Val accuracy: {val_acc:.4f} | "
            f"Val macro-F1: {val_f1:.4f}"
        )

    # Restore best validation checkpoint
    model.load_state_dict(best_state)

    test_loss, test_acc, test_f1 = evaluate_har_ablation(
        model,
        test_loader,
        criterion
    )

    result = {
        "problem": "P1",
        "model": f"HAR-{model_type}-h{hidden_size}",
        "seed": seed,
        "seq_len": np.nan,
        "test_acc": test_acc,
        "macro_f1": test_f1,
        "test_mse": np.nan,
        "params": params,
        "sec_per_epoch": np.mean(epoch_times),
        "val_loss_nats": np.nan,
        "bits_per_char": np.nan,
        "notes": (
            f"hidden_size={hidden_size}; "
            f"best_epoch={best_epoch}; "
            f"best_val_macro_f1={best_val_f1:.6f}"
        )
    }

    print(
        f"\nCompleted {model_type}, hidden={hidden_size}, seed={seed}"
    )
    print(f"Best epoch: {best_epoch}")
    print(f"Parameters: {params:,}")
    print(f"Test accuracy: {test_acc:.4f}")
    print(f"Test macro-F1: {test_f1:.4f}")
    print(f"Average seconds/epoch: {np.mean(epoch_times):.4f}")

    return result

# 7. Run all 24 experiments
ablation_results = []

for model_type in ["LSTM", "GRU"]:

    for hidden_size in HIDDEN_SIZES:

        print("\n" + "=" * 65)
        print(f"MODEL: {model_type} | HIDDEN SIZE: {hidden_size}")
        print("=" * 65)

        for seed in HAR_SEEDS:

            print(f"\nSeed: {seed}")

            result = train_har_ablation(
                model_type=model_type,
                hidden_size=hidden_size,
                seed=seed
            )

            ablation_results.append(result)

# Store results in a DataFrame
ablation_df = pd.DataFrame(ablation_results)

print("\nAll experiments completed!")
display(ablation_df)
```

### 5(c):  Ablation models and helpers

```python
# Experiment configuration
T_ABLATION = 200
HIDDEN_SIZE = 64
SEEDS = [42, 123, 2026]
EPOCHS = 20
BATCH_SIZE = 128
LEARNING_RATE = 1e-3

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


class Task5cModel(nn.Module):
    def __init__(self, model_type, hidden_size=64,
                 forget_bias=None):
        super().__init__()

        self.model_type = model_type

        if model_type == "RNN":
            self.rnn = nn.RNN(
                input_size=2,
                hidden_size=hidden_size,
                batch_first=True
            )

        elif model_type == "LSTM":
            self.rnn = nn.LSTM(
                input_size=2,
                hidden_size=hidden_size,
                batch_first=True
            )

            # Set the total forget-gate bias to the requested value.
            if forget_bias is not None:
                with torch.no_grad():
                    self.rnn.bias_ih_l0[
                        hidden_size:2 * hidden_size
                    ].fill_(forget_bias)

                    self.rnn.bias_hh_l0[
                        hidden_size:2 * hidden_size
                    ].zero_()

        else:
            raise ValueError("Unsupported model type")

        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        output, hidden = self.rnn(x)

        if self.model_type == "LSTM":
            h_n, c_n = hidden
        else:
            h_n = hidden

        final_hidden = h_n[-1]
        return self.fc(final_hidden).squeeze(-1)


def make_adding_loaders_for_5c(data, seed):
    train_dataset = TensorDataset(
        torch.tensor(data["X_train"], dtype=torch.float32),
        torch.tensor(data["y_train"], dtype=torch.float32)
    )

    val_dataset = TensorDataset(
        torch.tensor(data["X_val"], dtype=torch.float32),
        torch.tensor(data["y_val"], dtype=torch.float32)
    )

    test_dataset = TensorDataset(
        torch.tensor(data["X_test"], dtype=torch.float32),
        torch.tensor(data["y_test"], dtype=torch.float32)
    )

    generator = torch.Generator()
    generator.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        generator=generator
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False
    )

    return train_loader, val_loader, test_loader


def evaluate_mse(model, loader):
    model.eval()
    total_squared_error = 0
    total_samples = 0

    with torch.no_grad():
        for xb, yb in loader:
            xb = xb.to(device)
            yb = yb.to(device)

            predictions = model(xb)
            loss_sum = nn.functional.mse_loss(
                predictions, yb, reduction="sum"
            )

            total_squared_error += loss_sum.item()
            total_samples += len(yb)

    return total_squared_error / total_samples


def train_task5c(model, train_loader, val_loader,
                 clipping=None):
    model = model.to(device)
    optimizer = torch.optim.Adam(
        model.parameters(), lr=LEARNING_RATE
    )
    criterion = nn.MSELoss()

    best_val_loss = float("inf")
    best_state = None
    epoch_times = []

    for epoch in range(EPOCHS):
        model.train()

        if device.type == "cuda":
            torch.cuda.synchronize()

        start_time = time.perf_counter()

        for xb, yb in train_loader:
            xb = xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            predictions = model(xb)
            loss = criterion(predictions, yb)

            loss.backward()

            # Apply clipping only for the clipping experiment.
            if clipping is not None:
                nn.utils.clip_grad_norm_(
                    model.parameters(),
                    max_norm=clipping
                )

            optimizer.step()

        if device.type == "cuda":
            torch.cuda.synchronize()

        epoch_times.append(time.perf_counter() - start_time)

        val_loss = evaluate_mse(model, val_loader)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = copy.deepcopy(model.state_dict())

    model.load_state_dict(best_state)

    return model, best_val_loss, np.mean(epoch_times)


print("Task 5(c) model and helper definitions ready.")
```

##### Run both ablations across all three seeds

```python
data_200 = adding_data[T_ABLATION]
task5c_results = []

# A. RNN: gradient clipping vs no clipping
rnn_configs = [
    ("RNN-clip-none", None),
    ("RNN-clip-1.0", 1.0)
]

for model_name, clipping_value in rnn_configs:
    for seed in SEEDS:
        print(f"\n{model_name} | T={T_ABLATION} | seed={seed}")

        set_seed(seed)

        train_loader, val_loader, test_loader = (
            make_adding_loaders_for_5c(data_200, seed)
        )

        model = Task5cModel(
            model_type="RNN",
            hidden_size=HIDDEN_SIZE
        )

        params = sum(
            p.numel() for p in model.parameters()
            if p.requires_grad
        )

        model, best_val, sec_per_epoch = train_task5c(
            model,
            train_loader,
            val_loader,
            clipping=clipping_value
        )

        test_mse = evaluate_mse(model, test_loader)

        task5c_results.append({
            "problem": "P1",
            "model": model_name,
            "seed": seed,
            "seq_len": T_ABLATION,
            "test_acc": np.nan,
            "macro_f1": np.nan,
            "test_mse": test_mse,
            "params": params,
            "sec_per_epoch": sec_per_epoch,
            "val_loss_nats": np.nan,
            "bits_per_char": np.nan,
            "notes": (
                f"Task5c gradient clipping={clipping_value}; "
                f"best_val_mse={best_val:.6f}"
            )
        })

        print(
            f"Best val MSE: {best_val:.6f} | "
            f"Test MSE: {test_mse:.6f} | "
            f"Time/epoch: {sec_per_epoch:.3f}s"
        )

# B. LSTM: forget-gate bias 0 vs 1
lstm_configs = [
    ("LSTM-forget-bias-0", 0),
    ("LSTM-forget-bias-1", 1)
]

for model_name, forget_bias in lstm_configs:
    for seed in SEEDS:
        print(f"\n{model_name} | T={T_ABLATION} | seed={seed}")

        set_seed(seed)

        train_loader, val_loader, test_loader = (
            make_adding_loaders_for_5c(data_200, seed)
        )

        model = Task5cModel(
            model_type="LSTM",
            hidden_size=HIDDEN_SIZE,
            forget_bias=forget_bias
        )

        params = sum(
            p.numel() for p in model.parameters()
            if p.requires_grad
        )

        model, best_val, sec_per_epoch = train_task5c(
            model,
            train_loader,
            val_loader,
            clipping=None
        )

        test_mse = evaluate_mse(model, test_loader)

        task5c_results.append({
            "problem": "P1",
            "model": model_name,
            "seed": seed,
            "seq_len": T_ABLATION,
            "test_acc": np.nan,
            "macro_f1": np.nan,
            "test_mse": test_mse,
            "params": params,
            "sec_per_epoch": sec_per_epoch,
            "val_loss_nats": np.nan,
            "bits_per_char": np.nan,
            "notes": (
                f"Task5c forget_gate_bias={forget_bias}; "
                f"best_val_mse={best_val:.6f}"
            )
        })

        print(
            f"Best val MSE: {best_val:.6f} | "
            f"Test MSE: {test_mse:.6f} | "
            f"Time/epoch: {sec_per_epoch:.3f}s"
        )

print("\nAll Task 5(c) experiments completed.")
```

### Task 5(d): Overall Analysis and Discussion

## 5(d) Overall Analysis and Discussion

The experiments highlight the trade-off between recurrent model capacity, computational complexity, and the ability to learn long-term dependencies.

**Gradient behaviour and sequence length.** The gradient-flow experiment at \(T=100\) showed that the gradient norm at the earliest hidden state was effectively zero, while the final hidden state had a substantially larger gradient norm (0.00730 before training and 0.00267 after training). This illustrates the difficulty of propagating learning signals across many recurrent steps. On the adding problem, GRU performed best at sequence lengths 50, 100, and 200, solving all three runs at each length. LSTM solved two of three runs at \(T=50\), but neither LSTM nor vanilla RNN solved the task consistently at longer lengths. At \(T=400\), none of the models achieved the required test MSE below 0.05. Their performance approached the baseline, suggesting that learning the dependency between distant marked values becomes increasingly difficult as sequence length grows.

**Effect of hidden size.** Increasing LSTM hidden size from 16 to 128 improved mean test macro-F1 from 0.7431 to 0.8889, but increased the parameter count from 1,830 to 71,942. GRU achieved stronger performance with fewer parameters. GRU with 64 hidden units obtained the highest mean macro-F1 of 0.8937 (14,790 parameters), while GRU with 32 units achieved 0.8906 using only 4,326 parameters. Increasing GRU size to 128 did not provide further improvement. Thus, larger hidden representations do not necessarily justify their additional computational cost.

**Ablation findings.** At \(T=200\), gradient clipping with a maximum norm of 1.0 did not improve vanilla RNN performance: mean test MSE was 0.168652 with clipping versus 0.168256 without clipping. Initialising the LSTM forget-gate bias to 1 produced a marginally lower mean test MSE (0.168626) than initialising it to 0 (0.168740). However, these differences were small, and none of the 12 runs met the solved criterion. These results indicate that the tested interventions alone were insufficient to overcome the long-sequence learning challenge.

**Wearable-device recommendation.** For human activity recognition, GRU with 32 hidden units is a practical deployment candidate when model size and resource efficiency are important. It achieved a mean macro-F1 of 0.8906 with 4,326 parameters, close to the 0.8937 achieved by GRU with 64 units using 14,790 parameters. If maximum classification performance is prioritised over model size, GRU-64 is preferable. Actual deployment should additionally evaluate inference latency, memory consumption, and energy usage on the target wearable hardware.

Overall, the experiments show that GRU provides a favourable balance of classification performance and parameter efficiency for this HAR task, while all three recurrent architectures face increasing difficulty on the adding problem as sequence length grows.
